Escribe un programa que:

* Pida al usuario cuántos proyectos hay en una empresa.

* Para cada proyecto pida: nombre, inversión inicial y rentabilidad esperada (%).

* Guarde cada proyecto como una lista anidada [nombre, inversion, rentabilidad] dentro de una lista principal.

Calcule:

* El total invertido en todos los proyectos.

* La rentabilidad promedio.

* Cuántos proyectos tienen rentabilidad > 10%.

* La lista de nombres de esos proyectos.

* Imprima un resumen final.
(Usa variables, ciclos, listas anidadas, condicionales y funciones para calcular totales y promedios).

```python
import numpy as np
```

```python
n = int(input("Cuántos proyectos hay en una empresa? "))
lista =[]

for i in range(n):
  print(f"Proyecto {i+1}")
  nombre = input("Nombre: ")
  i_i = float(input("Inversión inicial: $"))
  r_e = float(input("Rentabilidad esperada (en decimal): "))

  proy = [nombre, i_i, r_e]
  lista.append(proy)

lista
```

```python
i_t = sum(lista[j][1] for j in range(n))
r_p = sum(lista[j][2] for j in range(n)) / n

c = 0
name = []

for j in range(n):
    if lista[j][2] > 0.10:
        c += 1
        name.append(lista[j][0])

print(f"La inversión total es de ${i_t:,.2f}")
print(f"La rentabilidad promedio es de {r_p*100:.2f}%")
print(f"Hay {c} proyectos con una rentabilidad mayor al 10% que son: {name}")
```

```python
import random as rd
n = 2000
calif = [rd.randint(0,100) for i in range(n)]
clave = [chr(rd.randint(33,126))+chr(rd.randint(33,126))+chr(rd.randint(33,126))+chr(rd.randint(33,126)) for i in range(n)]
```

```python
for i in range(n):
  print(f"{i+1}. {clave[i]} | {calif[i]}")
```

```python
cb = min(calif)
ca = max(calif)
pc = sum(calif) / n
cien = calif.count(100)

conteo = []
for i in range(min(calif),max(calif)+1):
  a = calif.count(i)
  conteo.append(a)
m = max(conteo)
i = conteo.index(m)

print(f"Calificación mínima: {cb}.")
print(f"Calificación máxima: {ca}.")
print(f"Promedio de calificaciones: {pc:.2f}.")
print(f"Calificación 100 aparece: {cien} veces.")
print(f'Moda de calificaciones: {i + min(calif)}.')
```

```python
cl_a, aprob = [],[]

for j in range(n):
  if calif[j] >= 60:
    aprob.append(calif[j])
    cl_a.append(clave[j])

prom = sum(aprob)/len(aprob)

print(f"Calificaciones aprobatorias hay: {len(aprob)}.")
print(f"Promedio de calificaciones aprobatorias: {prom:.2f}")
```

```python
cl_r, reprob = [],[]

for j in range(n):
  if calif[j] < 60:
    reprob.append(calif[j])
    cl_r.append(clave[j])

prom = sum(reprob)/len(reprob)

print(f"Calificaciones reprobatorias hay: {len(reprob)}.")
print(f"Promedio de calificaciones reprobatorias: {prom:.2f}")
```

```python
busqueda = input("Clave: ")
if busqueda in clave:
  i = clave.index(busqueda)
  print(f"{i+1}. {clave[i]}: {calif[i]}")
else:
  print(f"La clave {busqueda} no esta registrada.")
```
