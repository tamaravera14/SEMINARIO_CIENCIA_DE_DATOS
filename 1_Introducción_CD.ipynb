## Redes neuronales

```python
import tensorflow as tf
import numpy as np
celsius = np.array([-40, -10, 0, 8, 15, 22, 38], dtype=float)
fahrenheit = np.array([-40, 14, 32, 46, 59, 72, 100], dtype=float)
capa = tf.keras.layers.Dense(units=1, input_shape=[1])
modelo = tf.keras.Sequential([capa])
modelo.compile(optimizer=tf.keras.optimizers.Adam(0.1),loss='mean_squared_error')
```

```python
print("Comenzando entrenamiento...")
historial = modelo.fit(celsius, fahrenheit, epochs=1000, verbose=False)
print("Modelo entrenado!")
```

```python
import matplotlib.pyplot as plt
plt.xlabel("# Epoca")
plt.ylabel("Magnitud de pérdida")
plt.plot(historial.history["loss"])
```

```python
print("Hagamos una predicción!")
resultado = modelo.predict([[25]])
print("El resultado es ",resultado," fahrenheit!")
```

## Adivina el número

```python
#Juego Adivina el número
import random

intentosRealizados = 0

miNombre = input("¡Hola! ¿Cómo te llamas?")

número = random.randint(1, 20)
print("Bueno, ", miNombre ,", estoy pensando en un número entre 1 y 20.")

while intentosRealizados < 6:
  estimación = int(input("Intenta adivinar."))
  intentosRealizados = intentosRealizados + 1

  if estimación < número:
    print("Tu estimación es muy baja.")
  if estimación > número:
    print("Tu estimación es muy alta.")
  if estimación == número:
    break

if estimación == número:
  print("¡Buen trabajo, ", miNombre,  "! ¡Has adivinado mi número en ",intentosRealizados,"intentos!")

if estimación != número:
    print("Pues no. El número que estaba pensando era ", número)
```

```python
import textblob
from textblob import TextBlob
```

 * TextBlob:

    * **Analisis de sentimientos:** Evalúa si un texto es positivo, negativo (polaridad) u objetivo/subjetivo (subjetividad).
    * **Tokenización:** Divide un texto en palabras o frases individuales.
    * **Etiquetado gramatical (POS):** Identifica el rol de cada palabra (sustantivo, verbo, adjetivo, etc.).
    * **Traducción y corrección:** Permite traducir textos a otros idiomas o corregir errores ortográficos.

```python
texto = "I feel angry"
t = TextBlob(texto)
print(t.sentiment)
```

* **Polaridad (0.0)**: Mide qué tan positivo o negativo es el tono.
  * El rango va de -1.0 (muy negativo) a 1.0 (muy positivo).
  * Un valor de 0.0 indica que no muestra emociones a favor ni en contra.

* **Subjetividad (0.0)**: Mide si el texto es una opinión personal o un hecho.
  * El rango va de 0.0 (completamente objetivo) a 1.0 (completamente subjetivo).
  * Un valor de 0.0 significa que el texto presenta información factual sin opiniones

* *EJERCICIO*

Utilizar la fórmula de interes simple:

$$K_f = K_o ( 1 + i )^t$$

* capital_final ($K_f$) = $ 67, 275.00.

* capital_inicial ($K_o$) = $ 10, 000.00.

* TLR ($i$)= 0.10

* Tiempo ($t$)= 20 años

```python
capital_inicial = 10000
TLR = 0.6
t = 20

capital_final = capital_inicial * ( 1 + TLR)**t
print(f"Un capital de ${capital_inicial: ,} al {TLR*100: .0f}% en {t} años da ${capital_final: ,.2f}")
```

* Interes simple: la tasa de interes se cobra sobre el capital inicial siempre.
* Interes compuesto: Acumulado del interes generado del inicial y se hace como interes sobre interes de la nueva inversión.

## CICLO FOR

Función que ejecuta un proceso cierta cantidad de veces.

**for** i **in range**(numero de veces a repetir):
* El iterador inicia en **CERO** y termina en **n-1**.
* Para el for es (inicio, fin, paso)

```python
for i in range(10):
  print("Viva México!")
print("Fin")
```

```python
nombre = input("Dime tu nombre:").title()
for i in range(5):
  print(f"Hola {nombre}.")
```

```python
for i in range(5):
  nombre = input("Dime tu nombre:").title()
  print(f"Hola {nombre}.")
```

```python
# Mostrar numeros pares

for i in range(0,101,2):
  print(i)
```

```python
# se vende en 5 pzs por caja, el minimo de venta es de 10 cajas
# La capacidad maxima de producir es 2000 pzs (400 cajas)
# de 10 a 49 cajas el precio es de 7 por pz
# de 50 a 99 cajas del precio es de 6
# de 100 en adelante es de 5
# simular las ventas mostrando precio y total a pagar

for c in range(10,401):
  if 10 <= c < 50:
    precio = c * 5 * 7
    print(f"El precio a pagar es de ${precio:,.2f} en $7 cada pieza para {c} cajas.")
  elif 50 <= c < 100:
    precio = c * 5 * 6
    print(f"El precio a pagar es de ${precio:,.2f} en $6 cada pieza para {c} cajas.")
  elif 100 <= c <= 400:
    precio = c * 5 * 5
    print(f"El precio a pagar es de ${precio:,.2f} en $5 cada pieza para {c} cajas.")

```
