Objetivo:

Contexto:

Se desea establecer un sistema de control de Capital humano dentro de una empresa.

El equipo debe crear un programa en Python que:
* Permita ingresar la información necesaria para la administración del personal. Debe contener:

Clave - Nombre - Edad - Telefono - Antigüedad - Sueldo - Inasistencias

* Agregar 5 trabajadores, donde el usuario proporciona la información, la clave inicia con a mayúscula seguida de tres cifras, el teléfono que se crea mediante números aleatorios iniciando con 55 y con una extensión de 10 dígitos.

* Una vez ingresados 5 trabajadores desplegar un menú en en cual contenga las siguientes opciones:

  * Agregar otro trabajador y seguir preguntando si desea agregar más.
  * Buscar un trabajador mediante la clave, nombre o índice y mostrar toda su información, dando la opción de eliminar o editar cualquiera de sus datos.
  * Total de de inasistencias de la empresa
  * Pago total de la nómina
  * Sueldo promedio
  * Edad promedio
  * Mostrar el listado completo en forma descendiente por clave
  * Mostrar el nombre de todos los trabajadores que ganan mas de 15,000 pesos
  * Mostrar el nombre de todos los trabajadores que ganan menos de 5,000 pesos
  * Notificar a los trabajadores sin inasistencias que recibirán un bono de 3,000 pesos e indicar cual será sueldo mensual
  * Notificar a los trabajadores con 4 o más inasistencias que se encuentran en riesgo de perder el empleo.
  * El programa si ejecutándose hasta que el usuario diga que no desea realizar más operaciones.

```python
# Crear telefono y clave
import random as rd

empleados = []
n = 5

for i in range(n):
  clave = 'A'+ str(rd.randint(100,999))
  nombre = input("Nombre: ").title()
  edad = int(input("Edad: "))
  telefono = '55'+str(rd.randint(10000000,99999999))
  antiguedad = int(input("Antigüedad en años: "))
  sueldo = float(input("Sueldo: $"))
  inasistencias = int(input("Inasistencias: "))

  trabajador = [clave,nombre,edad,telefono,antiguedad,sueldo,inasistencias]
  empleados.append(trabajador)
```

* Una vez ingresados 5 trabajadores desplegar un menú en en cual contenga las siguientes opciones:

  * Agregar otro trabajador y seguir preguntando si desea agregar más.
  * Buscar un trabajador mediante la clave, nombre o índice y mostrar toda su información, dando la opción de eliminar o editar cualquiera de sus datos.
  * Total de de inasistencias de la empresa
  * Pago total de la nómina
  * Sueldo promedio
  * Edad promedio
  * Mostrar el listado completo en forma descendiente por clave
  * Mostrar el nombre de todos los trabajadores que ganan mas de 15,000 pesos
  * Mostrar el nombre de todos los trabajadores que ganan menos de 5,000 pesos
  * Notificar a los trabajadores sin inasistencias que recibirán un bono de 3,000 pesos e indicar cual será sueldo mensual
  * Notificar a los trabajadores con 4 o más inasistencias que se encuentran en riesgo de perder el empleo.
  * El programa si ejecutándose hasta que el usuario diga que no desea realizar más operaciones.

```python
def uno():
  r = 1
  while r == 1:
    clave = 'A'+ str(rd.randint(100,999))
    nombre = input("Nombre: ")
    edad = int(input("Edad: "))
    telefono = '55'+str(rd.randint(10000000,99999999))
    antiguedad = int(input("Antigüedad en años: "))
    sueldo = float(input("Sueldo: $"))
    inasistencias = int(input("Inasistencias: "))

    trabajador = [clave,nombre,edad,telefono,antiguedad,sueldo,inasistencias]
    empleados.append(trabajador)

    r = input("Agregar otro trabajador? (si/no) ").lower()
    if r == "si":
      r = 1
    else:
      r = 0
```

```python
def dos():
  decision = input("Lo buscara por (clave, nombre o índice) solo elegir una: ").lower()
  if decision == "clave":
    buscar = input("Clave: ").upper()
    for i in empleados:
      if i[0] == buscar:
        print(i)
        b_e = input("Lo quieres borrar o editar? ").lower()
        if b_e == "borrar":
          empleados.remove(i)
        else:
          cl = int(input("qué posición quieres editar? "))
          if cl == 0 or cl == 1 or cl ==3:
            i[cl] = input("Nuevo dato: ")
          elif cl == 5:
            i[cl] = float(input("Nuevo dato: "))
          else:
            i[cl] = int(input("Nuevo dato: "))
        break
    else:
      print("Clave invalida.")

  elif decision == "nombre":
    buscar = input("Nombre: ").title()
    for i in empleados:
      if i[1] == buscar:
        print(i)
        b_e = input("Lo quieres borrar o editar? ").lower()
        if b_e == "borrar":
          empleados.remove(i)
        else:
          cl = int(input("qué posición quieres editar? "))
          if cl == 0 or cl == 1 or cl ==3:
            i[cl] = input("Nuevo dato: ")
          elif cl == 5:
            i[cl] = float(input("Nuevo dato: "))
          else:
            i[cl] = int(input("Nuevo dato: "))
        break
    else:
      print("Nombre invalido.")

  elif decision == "indice":
    buscar = int(input("Índice: "))
    if buscar >= 0 or buscar <= len(empleados) :
      print(empleados[buscar])
      b_e = input("Lo quieres borrar o editar? ").lower()
      if b_e == "borrar":
        empleados.remove(buscar)
      else:
        cl = int(input("qué posición quieres editar? "))
        if cl == 0 or cl == 1 or cl ==3:
          empleados[buscar][cl] = input("Nuevo dato: ")
        elif cl == 5:
          empleados[buscar][cl] = float(input("Nuevo dato: "))
        else:
          empleados[buscar][cl] = int(input("Nuevo dato: "))

    else:
      print("Indice invalido.")

  else:
    print("No es una opcion de busqueda.")


print(empleados)
```

```python
def tres():
  total_i = 0
  for i in empleados:
    total_i += i[4]
  print(f"Total de inasistencias: {total_i}")
```

```python
def cuatro():
  total_nomina = 0
  for i in empleados:
    total_nomina += i[5]
  print(f"Total de nomina: ${total_nomina:,.2f}")
```

```python
def cinco():
  total_nomina = 0
  for i in empleados:
    total_nomina += i[5]
    s_prom= total_nomina/len(empleados)
  print(f"Sueldo promedio: ${s_prom:,.2f}")
```

```python
def seis():
  total_edad = 0
  for i in empleados:
    total_edad += i[2]
    e_prom= total_edad/len(empleados)
  print(f"Edad promedio: {e_prom}")
```

```python
def siete():
  clav = []
  for i in empleados:
    clav.append(i[0])
  clav.sort(reverse = True)
  #print(clav)

  for i in range(len(clav)):
    for j in range(len(clav)):
      if clav[i] in empleados[j]:
        print(empleados[j])
```

```python
def ocho():
    for i in empleados:
        if i[5] > 15000:
            print(f"{i[1]} tu sueldo es mayor a $15,000, ganas ${i[5]:,.2f}")
```

```python
def nueve():
    for i in empleados:
        if i[5] < 5000:
            print(f"{i[1]} tu sueldo es menor a $5,000, ganas ${i[5]:,.2f}")
```

```python
def diez():
  bono=0
  for i in empleados:
      if i[6] == 0:
          print(f"{i[1]} por no haber faltado se te dará un bono de $3,000, ahora ganas ${i[5]+3000:,.2f}.")
```

```python
def once():
  print("Estos empleados estan en riesgo de perder el empleo por inasistencia:")
  for i in empleados:
    if i[6]>=4:
      print(f"{i[1]} en riesgo, tienes {i[6]} inasistencias.")
```

```python
def menu():
  r = int(input("Elige una opción: "))
  while r < 1 or r > 11:
    r = int(input("Elige una opción válida: "))

  if r == 1:
    uno()
  elif r == 2:
    dos()
  elif r == 3:
    tres()
  elif r == 4:
    cuatro()
  elif r == 5:
    cinco()
  elif r == 6:
    seis()
  elif r == 7:
    siete()
  elif r == 8:
    ocho()
  elif r == 9:
    nueve()
  elif r == 10:
    diez()
  elif r == 11:
    once()
```
