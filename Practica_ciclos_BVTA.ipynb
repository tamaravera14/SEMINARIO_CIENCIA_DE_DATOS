1. Recibir un dato y elevarlo al cuadrado, repetirlo hasta que el usuario diga que ya no desea hacerlo.

```python
n = float(input("Dame un número: "))
print(f"Tu número {n} elevado al cuadrado es {n**2}.")
r = input("Deseas agregar otro? (si/no) ").lower()

while r == "si":
  n = float(input("Dame un número: "))
  print(f"Tu número {n} elevado al cuadrado es {n**2}.")
  r = input("Deseas agregar otro? (si/no) ").lower()

print("Programa finalizado.")
```

8. Escriba un programa que indique al usuario: "Cantidad a abonar?", el usuario ingresa una cantidad, esto se repite hasta que las sumas de las cantidades abonadas superan los $100,000. El programa no puede recibir cantidades negativas.


```python
s = 0

while True:
  c = float(input("Cantidad a abonar? $"))
  if c < 0:
    print("No cantidades negativas.")
    True
  else:
    True
    s += c
    if s >= 100000:
      print(f"Felicidades has abonado ${s:,.2f}.")
      break
```

9. Escriba un programa con una calculadora básica con las 6 operaciones básicas. El una vez que el usuario efectúa una operación, el programa pregunta si desea calcular otra operación o quiere finalizar el programa.

```python
while True:
  r = input("Deseas continuar? (si/no) ").lower()

  if r == "si":

    a = float(input('Dime un número: '))
    b = float(input('Dime un número: '))

    op = input("¿Potencia, raiz, multiplicación, división, suma o resta?").lower()

    if op == "suma":
      print(a+b)
    elif op == "resta":
      print(a-b)
    elif op == "division":
      print(a/b)
    elif op == "multiplicacion":
      print(a*b)
    elif op == "potencia":
      print(a**b)
    elif op == "raiz":
      print(a**(1/b))
    else:
      print("Operacion no valida.")
  else:
    print("Programa Finalizado")
    break
```

10. El mismo programa que el anterior pero antes de preguntar si desea continuar o finalizar, preguntará si desea realizar nuevamente la misma operación. El programa contiene dos whiles, uno general y otro específico por operación.

```python
continuar = "si"

while continuar == "si":

    op = input("¿Potencia, raiz, multiplicacion, division, suma o resta? ").lower()
    repetir = "si"

    while repetir == "si":

        a = float(input("Dime un número: "))
        b = float(input("Dime otro número: "))

        if op == "suma":
            print("Resultado:", a + b)
        elif op == "resta":
            print("Resultado:", a - b)
        elif op == "multiplicacion":
            print("Resultado:", a * b)
        elif op == "division":
            if b != 0:
                print("Resultado:", a / b)
            else:
                print("No se puede dividir entre cero.")
        elif op == "potencia":
            print("Resultado:", a ** b)
        elif op == "raiz":
            print("Resultado:", a ** (1 / b))
        else:
            print("Operacion no valida.")

        repetir = input("¿Deseas realizar nuevamente la misma operacion? (si/no): ").lower()
    continuar = input("¿Deseas realizar otra operacion? (si/no): ").lower()

print("Programa finalizado.")
```
